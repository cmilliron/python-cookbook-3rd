## 2.1 Splitting Strings on Any of Multiple Delimiters
**Problem**
You need to split a string into fields, but the delimiters (and spacing around them) aren’t consistent throughout the string.

**Solution**
The *split()* method of string objects is really meant for very simple cases, and does not allow for multiple delimiters or account for possible whitespace around the delimiters.

````python
line = 'asdf fjdk; afed, fjek,asdf,      foo'
````

````python
import re
````

````python
re.split(r'[;,\s]\s*', line)
````

````python
line
````

````python
fields = re.split(r'(;|,|\s)\s*', line)
fields
````

````python
values = fields[::2]
delimiters = fields[1::2]
````

````python
values
````

````python
delimiters
````

````python
"".join(v+d for v,d in zip(values, delimiters))
````

````python
# if you don't want to the separator characters in the results, but need to use parentheses to group parts of the regular expression pattern, make sure you use a noncapture group, specific as (?:...)

re.split(r'(?:,|;|\s)\s*', line)
````

Normally, fnmatch() matches patterns using the same case-sensitivity rules as the system’s underlying filesystem (which varies based on operating system). For example:


```python

>>> fnmatch('foo.txt', '*.TXT')
False

>>> # On Windows
>>> fnmatch('foo.txt', '*.TXT')
True
>>>
```

## 2.6 Searching and Replacing Case-insensitive Text

**Problem**
YOu need to search for and possible replace text in a case-insensitive manner

**Solution**

````python
text = 'UPPER PYTHON, lower python, Mixed Python'
````

````python
re.findall('python', text, flags=re.IGNORECASE)
````

````python
## I can work with sub too
re.sub('python', 'snake', text, flags=re.IGNORECASE)
````

````python
def matchcase(word):
    def replace(m):
        text = m.group()
        if text.isupper():
            return word.upper()
        elif text.islower():
            return word.lower()
        elif text[0].isupper():
            return word.capitalize()
        else:
            return word
    return replace
````

````python
re.sub('python', matchcase('snake'), text, flags=re.IGNORECASE)
````

## 2.8 Writing a Regular Expression for Multiline Patterns

**Problem**
You're trying to match a block of text using a regular expression, but you need the match to span multiple lines.

**Solution**
this problem typically arises in patterns that use the dot (.) to match any character but forget to account for the fact that it doesn't match newlines.

````python
comment = re.compile(r'/\*(.*?)\*/')
````

````python
text1 = '/* this is a comment */'
text2 = '''/* this is a
              multiline comment */
        '''
````

````python
comment.findall(text1)
````

````python
comment.findall(text2)
````

````python
comment_2 = re.compile(r'/\*((?:.|\n)*?)\*/')
````

````python
comment_2.findall(text1)
````

````python
comment_2.findall(text2)
````

**Discussion**

The re.compile() function accepts a flag, re.DOTALL, which is useful here. It makes the . in a regular expression match all characters, including newlines. For example:

```python
comment = re.compile(r'/\*(.*?)\*/', re.DOTALL)
comment.findall(text2)
```

[' this is a\n              multiline comment ']

Using the re.DOTALL flag works fine for simple cases, but might be problematic if you’re working with extremely complicated patterns or a mix of separate regular expressions that have been combined together for the purpose of tokenizing, as described in Recipe 2.18. If given a choice, it’s usually better to define your regular expression pattern so that it works correctly without the need for extra flags.

````python
# One downside of format() and format_map() is that they do not deal gracefully with missing values.capitalize

# s.format(name="Guido") # no n
````

````python
class safesub(dict):
    def __missing__(self, key):
        return '{' + key + "}"
````

````python
# del n #make sure n is undefined
````

````python
s.format_map(safesub(vars()))
````

````python
import sys

def sub(text):
    return text.format_map(safesub(sys._getframe(1).f_locals))
````
