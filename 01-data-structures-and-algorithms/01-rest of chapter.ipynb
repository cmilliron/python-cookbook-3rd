### Problem

### Solution
To find out what the two dictionaries have in common, simply perform common set operations using the keys() or items() methods.

```python
a = {
   'x' : 1,
   'y' : 2,
   'z' : 3
}

b = {
   'w' : 10,
   'x' : 11,
   'y' : 2
}
```

```python
a.items() & b.items()
```

## 1.10. Removing Duplicates from a Sequence while Maintaining Order

### Problem

You want to eliminate the duplicate values in a sequence, but preserve the order of the remaining items.

### Solution

If the values in the sequence are hashable, the problem can be easily solved using a set and a generator.

```python
def dedupe(items):
    seen = set()
    for item in items: 
        if item not in seen: 
            yield item
            seen.add(item)
```

```python
a = [1, 5, 2, 1, 9, 1, 5, 10]
```

```python
list(dedupe(a))
```

```python
# hash(a)
```

```python
def dedupe_dict(items, key=None):
    seen = set()
    for item in items:
        val = item if key is None else key(item)
        if val not in seen:
            yield item
            seen.add(val)
```

```python
b = [ {'x':1, 'y':2}, {'x':1, 'y':3}, {'x':1, 'y':2}, {'x':2, 'y':4}]
```

```python
list(dedupe_dict(b, key=lambda d: (d['x'], d['y'])))
```

```python
list(dedupe_dict(b, key=lambda d: (d['x'])))
```

## 1.13. Sorting a List of Dictionaries by a Common Key

**Problem**:
You have a list of dictionaries and you would like to sort the entries according to one or more of the dictionary values.

**Solution**:
Sorting this type of structure is easy using the operator module’s itemgetter function.

```python
rows = [
    {'fname': 'Brian', 'lname': 'Jones', 'uid': 1003},
    {'fname': 'David', 'lname': 'Beazley', 'uid': 1002},
    {'fname': 'John', 'lname': 'Cleese', 'uid': 1001},
    {'fname': 'Big', 'lname': 'Jones', 'uid': 1004}
]
```

```python
from operator import itemgetter
```

```python
rows_by_fname = sorted(rows, key=itemgetter('fname'))
rows_by_uid = sorted(rows, key=itemgetter('uid'))
```

```python
rows_by_fname_lambda = sorted(rows, key=lambda r: r["fname"])
rows_by_lfname_lambda = sorted(rows, key=lambda r: (r['lname'], r['fname']))
```

## 1.14. Sorting Objects Without Native Comparison Support

**Problem:** You want to sort objects of the same class, but they don’t natively support comparison operations.

**Solution:** The built-in sorted() function takes a key argument that can be passed a callable that will return some value in the object that sorted will use to compare the objects.

```python
class User: 
    def __init__(self, user_id):
        self.user_id = user_id

    def __repr__(self):
        return "User({})".format(self.user_id)
```

```python
users = [User(23), User(3), User(99)]
users
```

```python
sorted(users, key=lambda u: u.user_id)
```

```python
from operator import attrgetter
```

```python
sorted(users, key=attrgetter('user_id'))
```

## 1.15. Grouping Records Together Based on a Field

**Problem:**
You have a sequence of dictionaries or instances and you want to iterate over the data in groups based on the value of a particular field, such as date.

**Solution:**
The itertools.groupby() function is particularly useful for grouping data together like this. 

```python
rows = [
    {'address': '5412 N CLARK', 'date': '07/01/2012'},
    {'address': '5148 N CLARK', 'date': '07/04/2012'},
    {'address': '5800 E 58TH', 'date': '07/02/2012'},
    {'address': '2122 N CLARK', 'date': '07/03/2012'},
    {'address': '5645 N RAVENSWOOD', 'date': '07/02/2012'},
    {'address': '1060 W ADDISON', 'date': '07/02/2012'},
    {'address': '4801 N BROADWAY', 'date': '07/01/2012'},
    {'address': '1039 W GRANVILLE', 'date': '07/04/2012'},
]
```

```python
from operator import itemgetter
from itertools import groupby
```

```python
# Sort by the desired filed first
rows.sort(key=itemgetter('date'))
```

```python
# Iterate in groups
for date, items in groupby(rows, key=itemgetter('date')):
    print(date)
    for i in items:
        print("     ", i)
```

 ## 1.17 Extracting a Subset of a Dictionary

**Problem**
You want to make a dictionary that is a subset of another dictionary.

**Solution**
This is easily accomplished using a dictionary comprehension.


```python
prices = {
   'ACME': 45.23,
   'AAPL': 612.78,
   'IBM': 205.55,
   'HPQ': 37.20,
   'FB': 10.75
}
```

```python
# Make a dictionary of all prices over 200
pl = { key:value for key, value in prices.items() if value > 200}
pl
```

```python
# Make a dictiona of tech stocks
tech_names = { 'AAPL', "IBM", "HPQ", "MSFT"}
p2 = { key:value for key, value in prices.items() if key in tech_names}
p2
```

```python
## Can be duplicated through te use of dict() and passing tuples
p3 = dict((key, value) for key, value in prices.items() if value > 200)
p3
```

```python
# dictionary comprehension if much faster and cleaner
```

```python
# other options
# much slower
p4 = { key: prices[key] for key in prices.keys() & tech_names}
p4
```
